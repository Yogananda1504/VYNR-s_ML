#### Z - Score Method  

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from scipy.stats import boxcox

# Read the CSV file
df = pd.read_csv("data.csv")

# Perform Box-Cox transformation
df["cg_boxcox"], lambda_cg = boxcox(df["cg"])
df["marks_boxcox"], lambda_marks = boxcox(df["marks"])

# Detect outliers using z-scores
threshold = 3
df["cg_zscores"] = (df["cg"] - df["cg"].mean()) / df["cg"].std()
df["marks_zscores"] = (df["marks"] - df["marks"].mean()) / df["marks"].std()

# Trimming outliers
trimmed_df = df[
    (df["cg_zscores"].abs() < threshold) &
    (df["marks_zscores"].abs() < threshold)
]

# Capping outliers
cg_lower = df["cg"].mean() - threshold * df["cg"].std()
cg_upper = df["cg"].mean() + threshold * df["cg"].std()
df["cg_capped"] = np.where(df["cg"] < cg_lower, cg_lower,
                           np.where(df["cg"] > cg_upper, cg_upper, df["cg"]))

marks_lower = df["marks"].mean() - threshold * df["marks"].std()
marks_upper = df["marks"].mean() + threshold * df["marks"].std()
df["marks_capped"] = np.where(df["marks"] < marks_lower, marks_lower,
                              np.where(df["marks"] > marks_upper, marks_upper, df["marks"]))

print(df.head())


```
