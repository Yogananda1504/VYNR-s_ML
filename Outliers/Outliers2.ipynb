# IQR Method for Outlier Detection

The **Interquartile Range (IQR)** is a measure of statistical dispersion and is used to identify outliers in a dataset. It is calculated as the difference between the third quartile (Q3) and the first quartile (Q1):
```
IQR = Q3 - Q1

```

### Steps to Apply the IQR Method:

1. **Calculate Quartiles:**
    - **Q1 (25th percentile):** The value below which 25% of the data falls.
    - **Q3 (75th percentile):** The value below which 75% of the data falls.

2. **Compute IQR:**
    ```
    IQR = 69.1743 - 63.5056 = 5.6686
    
    ```

3. **Determine Outlier Boundaries:**
    - **Lower Bound:** \( Q1 - 1.5 \times \text{IQR} = 63.5056 - (1.5 \times 5.6686) = 55.5003 \)
    - **Upper Bound:** \( Q3 + 1.5 \times \text{IQR} = 69.1743 + (1.5 \times 5.6686) = 77.1762 \)

4. **Identify Outliers:**
    - Any data point below the lower bound (55.5003) or above the upper bound (77.1762) is considered an outlier.

### Example Application:

Using the IQR method on the `Height` variable:

- **IQR:** 5.6686
- **Q1:** 63.5056
- **Q3:** 69.1743
- **Outliers Identified:**
  - Heights below 55.5003
  - Heights above 77.1762

This method helps in effectively identifying and handling outliers, ensuring more accurate and reliable statistical analysis.

# Implementation

The following code demonstrates how to calculate the IQR and identify outliers in the dataset:

The IQR method provides a straightforward and effective way to detect outliers, ensuring that subsequent data analysis is based on reliable and consistent data.



````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df= pd.read_csv('weight-height.csv')
df.head(10)
````

````python
# Perform Box plot on weight and height 
plt.figure(figsize=(10,10))
sns.boxplot(df['Height'])
plt.show()

# Get the 25th, 50th and 75th percentile of height
q1 = df['Height'].quantile(0.25)
q2 = df['Height'].quantile(0.50)
q3 = df['Height'].quantile(0.75)
IQR=q3-q1
# Extract the outliers where the heights by IQR outlier rule 

outlier_data = df[(df['Height'] > (q3 + 1.5 * IQR)) | (df['Height'] < (q1 - 1.5 * IQR))]
outlier_data.head(10) 
````
