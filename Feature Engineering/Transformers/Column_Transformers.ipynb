```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.impute import SimpleImputer

# Load the data
data = pd.read_csv('output.csv')
original_index = data.index

# Define categorical order
cough_categories = ['Low', 'Mild', 'High']

# Define the column transformer with columns as lists
preprocessor = ColumnTransformer(
    transformers=[
        ('age_imputer', SimpleImputer(strategy='mean'), ['Age']),  # Changed to list
        ('fever_imputer', SimpleImputer(strategy='mean'), ['FeverTemp']),  # Changed to list
        ('onehot', OneHotEncoder(drop='first', sparse_output=False), ['Gender', 'City']),
        ('ordinal', OrdinalEncoder(categories=[cough_categories]), ['Cough']),  # Changed to list
        ('label', OrdinalEncoder(), ['HasCovid'])  # Changed to list
    ],
    remainder='passthrough',
    verbose_feature_names_out=True
)

# Apply transformations
data_transformed = preprocessor.fit_transform(data)
df = pd.DataFrame(
    data_transformed,
    index=original_index,
    columns=preprocessor.get_feature_names_out()
)

print("The data frame after applying the column transformer is : \n", df.sample(5))
```
