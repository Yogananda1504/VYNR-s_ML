```python
# This is the tutorial for the min-max scaling of the Normalization
# This technique is implemented when we know that the data is limited in a range 
# Centering around the minimum 
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

import pandas as pd


df= pd.read_csv('sample.csv')
print("This is the sample : \n",df.sample(5,ignore_index=True))
scaler=MinMaxScaler()
x=df[['feature1','feature2']]
y=df[['target']]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```
