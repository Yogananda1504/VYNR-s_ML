```python
from sklearn.preprocessing import MaxAbsScaler
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

print('\033[91m' + 'MaxAbs Scaling Usage:' + '\033[0m')
print('\033[91m' + '- Scales each feature by its maximum absolute value' + '\033[0m')
print('\033[91m' + '- Useful for sparse data with both positive and negative values' + '\033[0m')
print('\033[91m' + '- Preserves zero values in sparse data' + '\033[0m')
print('\033[91m' + '- Range of transformed data is [-1, 1]' + '\033[0m')

df=pd.read_csv('absScaling.csv')
x=df[['feature1','feature2']]
y=df['target']
scaler=MaxAbsScaler()
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
print('Before Scaling')
print(x_train)
print('After Scaling') 
print(pd.DataFrame(x_train_scaled,index=x_train.index,columns=x_train.columns))
print('\033[93m','After the scaling the data is converted into the numpy vector .So we should again convert it into the dataframe','\033[0m')

```
