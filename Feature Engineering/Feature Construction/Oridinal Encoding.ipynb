```python


'''
Note : The ordinal encoding is only applied to the input features and not the target variable.
       For the target variable we use the label encoding 
'''
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
import numpy as np
import pandas as pd

df = pd.read_csv('ordinaldata.csv')
x=df[['education_level','performance_level']]
y=df[['target']]
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)

# Define specific ordering for categories otherwise the encoder will assign the random orders which is not desired in ordinal encoding
education_categories = ['HighSchool', 'Bachelors', 'Masters', 'PhD']
performance_categories = ['Poor', 'Average', 'Good', 'Excellent']

# Initialize encoder with custom ordering
encoder = OrdinalEncoder(categories=[education_categories, performance_categories])
x_train_encoded = encoder.fit_transform(x_train)
x_test_encoded = encoder.transform(x_test)

# Label encoding for target variable
label_encoder = LabelEncoder()
'''
Oridinal Encoding is used when the categorical data is ordinal in nature . 
y_train.values.ravel() is a combination of two operations:
1. y_train.values:
    - Converts the pandas DataFrame/Series into a NumPy array
    - Removes the pandas index and column labels
    - Returns just the raw data values
2. ravel():
    - Flattens a multi-dimensional array into a 1-dimensional array
    - Returns a view of the original array when possible (memory efficient)
    - Required by LabelEncoder which expects 1D input
For example:
If y_train is:
      target
0    Yes
1    No
2    Yes
y_train.values returns:
[['Yes']
 ['No']
 ['Yes']]
y_train.values.ravel() returns:
['Yes' 'No' 'Yes']
This flattened 1D array format is required for proper label encoding.
'''
y_train_encoded = label_encoder.fit_transform(y_train.values.ravel())
y_test_encoded = label_encoder.transform(y_test.values.ravel())

# Print the encoded data to verify
print("Original training features:\n", x_train.head())
print("\nEncoded training features:\n", x_train_encoded[:5])
print("\nOriginal target values:\n", y_train.head())
print("\nEncoded target values:\n", y_train_encoded[:5])

# Display the custom mappings
print("\nEducation Level Mapping:")
for i, category in enumerate(education_categories):
    print(f"{category}: {i}")

print("\nPerformance Level Mapping:")
for i, category in enumerate(performance_categories):
    print(f"{category}: {i}")

print("\nTarget Mapping:")
print(dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_))))


 
```
