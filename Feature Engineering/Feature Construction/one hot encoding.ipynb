```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

def one_hot_encode_tutorial():
    # 1) Read the CSV data into a DataFrame
    df = pd.read_csv("oneHotEncodingData.CSV")
    
    # 2) One-hot encoding using pandas
    df_pandas_encoded = pd.get_dummies(df, columns=["Color", "Category"], prefix=["Color", "Category"], drop_first=True)
    print("Pandas get_dummies result:")
    print(df_pandas_encoded.head())

    # 3) One-hot encoding using scikit-learn
    encoder = OneHotEncoder(drop='first', sparse=False)
    # Fitting on the 'Color' and 'Category' columns
    encoded_features = encoder.fit_transform(df[["Color", "Category"]])
    feature_names = encoder.get_feature_names_out(["Color", "Category"])
    # Combine encoded features with the original DataFrame
    df_sklearn_encoded = pd.concat(
        [df["ID"], pd.DataFrame(encoded_features, columns=feature_names)],
        axis=1
    )
    print("\nScikit-learn OneHotEncoder result:")
    print(df_sklearn_encoded.head())

if __name__ == "__main__":
    one_hot_encode_tutorial()
```
