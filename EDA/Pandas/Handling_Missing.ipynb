```python
import pandas as pd 
import seaborn as sns
import numpy as np 

df= pd.read_csv('Iris.csv');
df.head()
df.describe() #Summary statistics
df.info() #Information about the data
df.isnull() ## Returns the boolean Data frame where True means the value is missing
df.isnull().sum() #Returns the number of missing values in each column

# Visualize missing values
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(df.isnull(), cbar=False)
plt.show()
```

```python
df_mean=pd.read_csv('sample.csv')
df_median=pd.read_csv('sample.csv')
df_mode=pd.read_csv('sample.csv')
#Performing the mean imputation on the df data frame
df_mean.fillna(df_mean.mean(), inplace=True)
print(df_mean)
#performing the median imputation on the df data frame
df_median.fillna(df_median.median(), inplace=True)
print(df_median)
#performing the mode imputation on the df data frame
df_mode.fillna(df_mode.mode(), inplace=True)
print(df_mode)
```

```python
df=pd.read_csv('sample.csv');
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(df.isnull(), cbar=False)
plt.show()
df.isnull().sum().plot(kind='bar')
plt.show()
print("This is before the imputation of the missing values : \n")
print(df)

#Performing the KNN imputation
from sklearn.impute import KNNImputer
imputer = KNNImputer()
df_imputed=pd.DataFrame(imputer.fit_transform(df),columns=df.columns)
print("This is after the imputation of the missing values : \n")
print(df_imputed)
```
