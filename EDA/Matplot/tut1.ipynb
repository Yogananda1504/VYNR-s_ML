```python
import matplotlib.pyplot as plt
import numpy as np

# Generate random data
x = np.random.rand(50)
y = np.random.rand(50)
colors = np.random.rand(50)
sizes = 1000 * np.random.rand(50)

# Create scatter plot
plt.scatter(x, y, c=colors)

# Add colorbar
plt.colorbar(label='Color Scale')

# Add title and labels
plt.title('Scatter Plot Example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

# Display the plot
plt.show()
```

```python
#Bar Chart
import matplotlib.pyplot as plt
categories=['A','B','C','D','E']
values=[7,3,12,5,2]
plt.bar(categories,values,color='pink')
plt.show()
```

```python
#Histogram
import matplotlib.pyplot as plt
import numpy as np

# Generate random data
data = np.random.randn(1000)

# Create histogram
plt.hist(data, bins=30, alpha=0.7, color='green', edgecolor='black')

# Add title and labels
plt.title('Histogram Example')
plt.xlabel('Value')
plt.ylabel('Frequency')

# Display the plot
plt.show()
```

```python
#Sub plots
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Create subplots: 1 row, 2 columns
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# First subplot
ax1.plot(x, y1, 'b-', label='Sine')
ax1.set_title('Sine Wave')
ax1.set_xlabel('x')
ax1.set_ylabel('sin(x)')
ax1.legend()
ax1.grid(True)

# Second subplot
ax2.plot(x, y2, 'r--', label='Cosine')
ax2.set_title('Cosine Wave')
ax2.set_xlabel('x')
ax2.set_ylabel('cos(x)')
ax2.legend()
ax2.grid(True)

# Adjust layout
plt.tight_layout()

# Display the plot
plt.show()
```
