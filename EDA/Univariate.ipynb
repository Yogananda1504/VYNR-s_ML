```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data =pd.read_csv('sample.csv')
#univariate analysis - The process of analysis a particular column depending upon the data type of the column (Numerical or Categorical data)

#Histogram 
plt.hist(data['age'],bins=20,color='blue',edgecolor='black')


#DistPlot with KDE
sns.displot(data['age'], bins=20, color='blue')

#boxplot
sns.boxplot(data['age'],color='blue') 

data['age'].describe()



```
