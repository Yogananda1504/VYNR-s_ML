```python
# Import necessary libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load an example dataset
data = sns.load_dataset('tips')

# Display the first few rows of the dataset
print(data.head())

# Basic Scatter Plot
plt.figure(figsize=(8, 6))
sns.scatterplot(x='total_bill', y='tip', data=data)
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()

# Basic Bar Plot for Size vs Total Bill
plt.figure(figsize=(8, 6))
sns.barplot(x='size', y='total_bill', data=data)
plt.title('Bar Plot of Size vs Total Bill')
plt.show()

# Basic Histogram
plt.figure(figsize=(8, 6))
sns.histplot(data['total_bill'], bins=20)
sns.kdeplot(data['total_bill'], color='r', lw=2)
plt.title('Histogram of Total Bill')
plt.show()

# Basic Box Plot
plt.figure(figsize=(8, 6))
sns.boxplot(x='day', y='total_bill', data=data)
plt.title('Box Plot of Total Bill by Day')
plt.show()

# Basic Bar Plot
plt.figure(figsize=(8, 6))
sns.barplot(x='day', y='total_bill', data=data)
plt.title('Bar Plot of Total Bill by Day')
plt.show()

# Basic Heatmap
plt.figure(figsize=(8, 6))
corr = data.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Heatmap of Correlation Matrix')
plt.show()

# Pair Plot
sns.pairplot(data)
plt.title('Pair Plot of the Dataset')
plt.show()

# Customizing plots
plt.figure(figsize=(8, 6))
sns.scatterplot(x='total_bill', y='tip', hue='time', style='time', size='size', data=data)
plt.title('Customized Scatter Plot')
plt.show()
```
