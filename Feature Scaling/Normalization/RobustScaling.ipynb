```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Create a dataset with 100 samples
n_samples = 100

# Generate features with different scales and some outliers
feature1 = np.random.normal(0, 1, n_samples)  # Normal distribution
feature2 = np.random.exponential(2, n_samples)  # Exponential distribution with outliers
feature3 = np.random.uniform(-10, 10, n_samples)  # Uniform distribution

# Add some extreme outliers
feature1[0] = 15
feature2[1] = 50
feature3[2] = 30

# Create a DataFrame
df = pd.DataFrame({
    'feature1': feature1,
    'feature2': feature2,
    'feature3': feature3
})

# Save to CSV
df.to_csv('sparse_data.csv', index=False)

# Display first few rows
print("First few rows of the created dataset:")
print(df.head())
```
