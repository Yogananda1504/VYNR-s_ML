```python
# MEAN Imputation in the case when probability density function is nearly normal
# MEDIAN Imputation in the case when probability density function is not normal

import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split

'''
Advantages : Simple and Easy to implement .
Disadvantages : i) Distorts the original distribution of the variable.
               ii) Possibility of extra outliers being generated.
               iii) Covariance of column with the other columns might get effected
               
When to Use : When the Data is missing and that to less than or equal to 5% of the total data 
'''

# Using the Titanic Dataset
df = pd.read_csv('../Datasets/Titanic.csv')
df=df[['Age','Fare','Parch','Survived']]


#  Checking the percentage of the missing values in each columns of the Dataset
missing_values = df.isnull().mean()*100
print(missing_values)

# Ingeneral if the missing data is more than that of the 5% then the simple imputation is not recommended
# Anyhow for the tutorial purposes we are doing here 


X = df[['Age', 'Fare', 'Parch']]
y = df['Survived']
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Calculate the mean and median of the Age and Fare columns
mean_age=x_train['Age'].mean()
median_age=x_train['Age'].median()

mean_fare=x_train['Fare'].mean()
median_fare=x_train['Fare'].median()

#Impute
x_train['Age_median'] = x_train['Age'].fillna(median_age)
x_train['Age_mean'] = x_train['Age'].fillna(mean_age)

x_train['Fare_median'] = x_train['Fare'].fillna(median_fare)
x_train['Fare_mean'] = x_train['Fare'].fillna(mean_fare)


# After performing the mean and median imputation we need to check the variance . 
'''
  The variance in the case of the mean and median imputation might change as most of the data which is missing is now filled with the mean or median hence the variance decreases
  This cause the change in the PD Curve of the data. 
  Huge variance after the imputation is a red flag denoting we should try other imputation method 
'''

print('Original Variance of Age :',x_train['Age'].var())
print('Variance after Median Imputation of age:',x_train['Age_median'].var())
print('Variance after Mean Imputation of age:',x_train['Age_mean'].var())

print('Original Variance of Fare :',x_train['Fare'].var())
print('Variance after Median Imputation of Fare:',x_train['Fare_median'].var())
print('Variance after Mean Imputation of Fare:',x_train['Fare_mean'].var())
```

#### As we can see that the  Variance of the Age after the imputation has changed in huge amount .  

```python
# Using the simple Imputer 
#  For the mean imputation use the mean and median imputation use the median and for the mode use the most_frequent
imputer = SimpleImputer(strategy='mean')
x_train_imputed = imputer.fit_transform(x_train)
x_train_imputed = pd.DataFrame(x_train_imputed, columns=x_train.columns)
print('Variance after Simple Imputer:',x_train_imputed['Age'].var())
print('Variance after Simple Imputer:',x_train_imputed['Fare'].var())
```
